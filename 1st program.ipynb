```python

import pandas as pd
import numpy as np


data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Date": ["2024-01-15", "15/02/2024", "March 10, 2024", "invalid_date"],
    "Sales": [200, 150, 300, 250]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
print("\nAfter Converting to Datetime:")
print(df)
df["Date"] = df["Date"].fillna(pd.Timestamp("2024-01-01"))
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Day_of_Week"] = df["Date"].dt.dayofweek
df["Is_Weekend"] = df["Day_of_Week"].apply(lambda x: 1 if x >= 5 else 0)
df["Quarter"] = df["Date"].dt.quarter
df["Month_Name"] = df["Date"].dt.month_name()
df["Week_of_Year"] = df["Date"].dt.isocalendar().week
df = pd.get_dummies(df, columns=["Month_Name"], drop_first=True)
df = df.drop(columns=["Date", "Name"])
print("\nFinal Preprocessed Data:")
print(df)
```
